### **Τύποι μεταβλητών**

Όλες οι γλώσσες προγραμματισμού (ή, τουλάχιστον, όσες κατέχει ο γράφων) αναγνωρίζουν 4 τύπους μεταβλητών, τους οποίους πρέπει να κατανοήσουμε καλά πριν προχωρήσουμε:

*  *Ακέραιοι* (``int`` από το integers): 0,1,2,-1...
*  *Δεκαδικοί* (``float``, από το floating point- κινητή υποδιαστολή): Το κομβικό να καταλάβουμε εδώ είναι πως ένας δεκαδικός που τυχαίνει να έχει δεκαδικό μέρος 0 είναι ουσιωδώς διαφορετικός από έναν ακέραιο! Το 3 δεν είναι το ίδιο με το 3.0!
*  *Κείμενο* (``str`` από το string, που με τη σειρά του σημαίνει "συμβολοσειρά"): Η Python πλέον υποστηρίζει, εκτός από λατινικούς χαρακτήρες και βασικά σύμβολα, και ελληνικούς χαρακτήρες, ϖ, Ϡ, ♠, ∫, ό,τι τραβάει η ψυχή σας. Τόσο για λόγους ταχύτητας όμως, όσο και για λόγους readability$^1$ και reproducability$^2$ συστήνεται η αποφυγή περίεργων χαρακτήρων όταν είναι δυνατόν.
*  *Λογικές* μεταβλητές (``bool`` από το boolean, που ο γράφων δε γνωρίζει αν έχει κάποιο άλλο νόημα): Πρακτικά, ``True`` ή ``False``. Όπως λοιπόν το 3 δεν είναι ίδιο με το 3.0, το ``True`` για την Python δεν είναι λέξη, είναι δήλωση ότι κάτι ισχύει!

$1$ Readability: Η εξίσωση $y=x$, όπως και να το κάνουμε, διαβάζεται πιο εύκολα από την ⨁=⨝.

$2$ Reproducability: Το να γράψω θ αντί theta είναι εύκολο για εμένα, αλλά τσάμπα κόπος για οποιονδήποτε δεν έχει ελληνικά στο πληκτρολόγιο. Παρ'όλα αυτά, στο συγκεκριμένο notebook θα βρείτε ελληνικά, ακριβώς επειδή προορίζεται μόνο για φοιτητές του ΕΚΠΑ.

### **Η εντολή `for`**

Μία από τις σημαντικότερες χρήσης μίας γλώσσας προγραμματισμού είναι ότι μπορούμε να περιγράψουμε μία διαδικασία και να δώσουμε τις κατάλληλες εντολές ώστε αυτή να επαναληφθεί πολλές φορές. Η θεμελιώδης εντολή με την οποία ορίζουμε μία τέτοια "λούπα" είναι η `for`.

```python
#Η "πλήρης" σύνταξη μίας λούπας for γίνεται ως εξής: for δείκτης in range (αρχή, τέλος, βήμα)
for i in range(0,20,2):#Το "τέλος" δε συμπεριλαμβάνεται! Αυτό δηλαδή πηγαίνει μέχρι το 18.
  print(i)            #Κατά τα γνωστά, μετά την : και για όσο γράφουμε σε εσοχή είμαστε "εντός" της λούπας. Ό,τι γράψουμε εδώ γίνεται σε κάθε βήμα.
```

```python
#Όταν γράφουμε απλώς range(τέλος), αρχίζει αυτόματα από το 0 και έχει βήμα 1
for i in range(20):
  print(i)
```

### **Η εντολή `while`**

Η `while` δουλεύει όπως η `for` αλλά, αντί να επικεντρώνεται στο πλήθος των επαναλήψεων, επικεντρώνεται σε μία συνθήκη. Όταν η συνθήκη πάψει να ικανοποιείται, είτε έχουν χρειαστεί 2 είτε 22 επαναλήψεις, η λούπα σταματά.

```python
x = 0
while x<10: #Το while ακολουθείται από τη συνθήκη
  x+=3
  print(x)  #Δε μας απασχόλησε καθόλου το πλήθος των επαναλήψεων!
```

> *  Εάν ο αριθμός δεν είναι θετικός, θα εμφανίζει ένα μήνυμα του στυλ "Σφάλμα, προσπαθήστε ξανά."
> *  Θα επαναλαμβάνει αυτή τη διαδικασία μέχρι να δοθεί θετικός αριθμός.
>
> Κάντε το ίδιο με `while`. Ποιος τρόπος συμφέρει σε αυτή την περίπτωση;

### **Γνωριμία με λίστες και πίνακες**

Μία λίστα στην Python είναι ένα σύνολο από στοιχεία, εντός [αγκυλών], που διαχωρίζονται με κόματα. Π.χ. η `A=[2 ,3.0, 'dog']` είναι μία λίστα. Ακριβώς επειδή τα στοιχεία τους δεν είναι απαραίτητα αριθμοί, οι λίστες δεν είναι μαθηματικοί πίνακες και πράξεις όπως η πρόσθεση δεν έχουν (ιδιαίτερο) νόημα.

```python
A=[2,3,4]
B=[5,6,7]
print(A+B)
```

Κατά παράδοση, στην πληροφορική μετράμε από το 0. Επομένως το πρώτο στοιχείο ενός πίνακα `A` είναι το `A[0]`, το δεύτερο το `A[1]` κλπ. Το τελευταίο στοιχείο μπορούμε να το γράψουμε και `A[-1]`, το προ-τελευταίο `A[-2]` κλπ.

Πώς όμως κατασκευάζουμε έναν μαθηματικό πίνακα, ή έστω κάτι ανάλογο της λίστας το οποίο να εμπίπτει στη μαθηματική λογική; Η "συνταγή" δίνεται από τη βιβλιοθήκη `numpy` και είναι `np.array(λίστα)`. Π.χ.

```python
A=np.array([1,2,3]) # Απευθείας
B=[4,5,6]
B=np.array(B) # Ή και σε 2 βήματα
print(A+B)
```


> **Άσκηση:** Ορίστε δύο αριθμητικούς πίνακες και εξερευνήστε τις πράξεις μεταξύ τους και μεταξύ των στοιχείων τους. Π.χ. `A - B`, `A[0]*B`, `A[0]*B[-1]`.

### **Λίστες (συνέχεια)**

Μπορούμε να δημιουργήσουμε δισδιάστατες λίστες ως εξής: δημιουργούμε μία λίστα, τα στοιχεία της οποίας είναι λίστες! Π.χ. `A=[[2,4,6], [7,5,3]]`. Εν συνεχεία, είναι απλό να κατασκευάσουμε έναν μαθηματικό πίνακα: `Β = np.array(A)` ή ακόμα και να γενικεύσουμε, κατασκευάζοντας τρισδιάστατες λίστες (λίστες, με στοιχεία λίστες, με στοιχεία λίστες), μαθηματικούς "κύβους" κ.ο.κ.

* Πώς αναφερόμαστε σε ένα συγκεκριμένο στοιείο ενός πίνακα;

> `Β[i][j]` το στοιχείο στην `i` γραμμή και `j` στήλη

```python
# Πώς παίρνουμε το μέγεθος μιας λίστας
A = [1, 2, 3, 4]
megethos = len(A) # συνάρτηση που παίρνει μια λίστα και επιστρέφει το μέγεθός της
print(len(A))
```

Τι θα βγάλει η συνάρτηση `len()` εάν της ταϊσουμε έναν πίνακα (μια δισδιάστατη λίστα); Στο ακόλουθο παράδειγμα της δίνουμε τον πίνακα $A=\begin{pmatrix} 1 & 2 \\ 2 & 3 \\ 6 & 0\end{pmatrix}$

```python
A = [[1, 2], [2, 3], [6, 0]]

print(len(A))    # επιστρέφει 3. Μας μετράει τις γραμμές!!!

print(len(A[0])) #τυπώνει πόσες στήλες έχει ο πίνακας
```

### **Γραφήματα στη Python**

Για τη δημιουργία γραφημάτων θα αξιοποιήσουμε τη βιβλιοθήκη `matplotlib`. Συγκεκριμένα, θα φορτώνουμε την `matplotlib.pyplot` δίνοντάς της συνήθως το "παρατσούκλι" `plt`.

Η διαδικασία που ακολουθούμε είναι: φτιάξε ένα κενό γράφημα $\rightarrow$ κατασκεύσασε ό,τι γραφική χρειάζεται $\rightarrow$ εμφάνισε ή/και αποθήκευσε το γράφημα.

Η εντολή δημιουργίας ενός γραφήματος είναι `plt.figure()`. Οι `()` υποδεικνύουν ότι θα μπορούσαμε να βάλουμε και κάποιο όρισμα εκεί, αλλά δεν απαιτείται.$^*$

Η εντολή με την οποία απεικονίζουμε κάτι είναι η `plt.plot(συντ/μένη x,υντ/μένη y, άλλα ορίσματα)`. Θα τη μελετήσουμε σε βάθος συντόμως.

Η εντολή με την οποία εμφανίζουμε το γράφημα είναι η `plt.show()`, ενώ η εντολή με την οποία το αποθηκεύουμε (και πρέπει να γράφεται *πριν* τη `show()`) είναι η `plt.savefig('όνομα αρχείου')`.

Τώρα, όσον αφορά τα άλλα ορίσματα της `plt.plot`:

$^*$ Το όρισμα αυτό θα μπορούσε π.χ. να είναι το μέγεθος, ως `figsize=(x,y)`.

* Η συμβολοσειρά (string) `"ο"` δηλώνει πώς τα δεδομένα θα ζωγραφιστούν
  ως γεμάτοι κύκλοι. Σαν επιλογές, μπορείτε να δοκιμάσετε και "+", "*", ".".

  <span style="color:olive">Σημείωση:</span> Εάν δε τοποθετήσουμε όρισμα για το σχήμα
  των δεδομένων που ζωγραφίζουμε, τότε η `matplotlib` θα ενώσει τα σημεία μας

### **Fit δεδομένων με πολυωνυμικές συναρτήσεις**

* `np.polyfit()` Ως ορίσματα λαμβάνει:\
1. Την αριθμητική λίστα με τις x συντεταγμένες.
2. Την αριθμητική λίστα με τις y συντεταγμένες.
3. Τον αριθμό της μεγαλύτερης δύναμης ή αλλιώς βαθμό του πολυωνύμου.

Επιστρέφει μια αριθμητική λίστα `np.array()` με τους βέλτιστους συντελεστές των δυνάμεων του $x$ από τη μεγαλύτερη προς τη μικρότερη δύναμη.

Π.χ. Για ένα πολυώνυμο δευτέρου βαθμού, $ax^2 + bx + c = 0$ επιστρέφει τη λίστα `[a, b, c]`.

### **Προσαρμογή ευθείας σε δεδομένα**

```python
X = np.array([2, 3, 6, 8, 13])
Y = np.array([5.5, 10, 15, 28, 36])

S = np.polyfit(X, Y, 1) # πολυώνυμο πρώτου βαθμού, άρα εξίσωση ευθείας y = ax + b

Z = S[0]*X + S[1] # s[0] -> a,  s[1] -> b.

plt.figure()

plt.plot(X, Y, "o", color="cornflowerblue") # αυτά είναι τα δεδομένα
plt.plot(X, Z, color="crimson") # αυτή είναι η σχεδίαση της ευθείας.

plt.show()

# Μπορώ να τυπώσω τις παραμέτρους ώστε να τις βλέπω
print("Βέλτιστη προσαρμογή ευθείας y = ax + b, για παραμέτρους:")
print("a = ", S[0])
print("b = ", S[1])
```

### **Προσαρμογή παρεμβολής σε δεδομένα**

```python
X = np.array([2, 3, 6, 8, 13])
Y = np.array([5.5, 10, 28, 50, 180])

S = np.polyfit(X, Y, 2) # πολυώνυμο δευτέρου βαθμού, άρα εξίσωση παραβολής y = ax^2 + bx + c

Z = S[0]*X**2 + S[1]*X + S[2] # s[0] -> a,  s[1] -> b,  s[2] -> c.

plt.figure()

plt.plot(X, Y, "o", color="cornflowerblue") # αυτά είναι τα δεδομένα
plt.plot(X, Z, color="crimson") # αυτή είναι η σχεδίαση της παραβολής.

plt.show()

# Μπορώ να τυπώσω τις παραμέτρους ώστε να τις βλέπω
print("Βέλτιστη προσαρμογή παραβολής y = ax^2 + bx + c, για παραμέτρους:")
print("a = ", S[0])
print("b = ", S[1])
print("c = ", S[2])
```

### **Μία γεύση από Monte Carlo**

Οι προσομοιώσεις Monte Carlo (από το ομώνυμο casino) έχουν την εξής γενική λογική: θέλοντας να λύσουμε ένα *ντετερμινιστικό* πρόβλημα (δηλαδή χωρίς τυχαιότητα- πρακτικά συνήθως θέλουμε να υπολογίσουμε ένα ολοκλήρωμα) εισάγουμε τυχαίους αριθμούς και, αν η λύση μας είναι σωστή, για μεγάλο πλήθος τυχαίων αριθμών αναπαράγουμε το μη ντετερμινιστικό αποτέλεσμα.

Π.χ. Αν γεμίσουμε ένα υποσύνολο $A$ του επιπέδου $xy$ με $N$ τυχαία σημεία, και ένα υποσύνολο αυτών, $n$, πέσει σε ένα μικρότερο υποσύνολο $B$ του επιπέδου, ισχύει

$$\frac{B}{A}=\frac{n}{N}$$

δηλαδή ο λόγος των επιφανειών ισούται με το λόγο των σημείων (εφόσον, φυσικά, $N\gg$). Το πόσα σημεία θεωρούνται "πολλά" προφανώς εξαρτάται από τους βαθμούς ελευθερίας, αλλά και την πολυπλοκότητα του προβλήματος.

Στο ακόλουθο παράδειγμα θα αξιοποιήσουμε τη μέθοδο Monte Carlo για να υπολογίσουμε τον αριθμό $\pi$. Η λογική είναι η εξής: έστω ένας κύκλος ακτίνας $R$. Αυτός είναι *εγγεγραμμένος* σε ένα τετράγωνο πλευράς $2R$. Τα εμβαδά τους είναι $\pi R^2$ και $(2R)^2$ αντίστοιχα, άρα ισχύει
$$\frac{4A_{\sf O}}{A_\square}=\pi$$

Μάλιστα, το αποτέλεσμα αυτό δεν εξαρτάται από την ακτίνα $R$, οπότε ας τη θεωρήσουμε $1$ χωρίς βλάβη της γενικότητας.

Όπως αιτιολογήσαμε νωρίτερα, για μεγάλα $N$ μπορούμε να προσεγγίσουμε το λόγο αυτό ως

$$\pi\approx\frac{4n}{N}$$

Η συνθήκη που εξασφαλίζει ότι ένα σημείο βρίσκεται εντός του μοναδιαίου κύκλου (μετράει δηλαδή στο $n$) είναι $x^2+y^2<1$.

Ο παραπάνω συλλογισμός, που δεν αλλοιώνεται εάν περιοριστούμε στο άνω δεξιά τεταρτημόριο, υλοποιείται στον ακόλουθο κώδικα:

```python
N=10000 # Αριθμός ψευδοτυχαίων σημείων που θα γεννήσουμε στην τετραγωνική επιφάνεια
n=0     # αρχικοποίηση. Σημεία που θα ικανοποιούν τη συνθήκη να βρίσκονται κάτω απ το
        # τόξο του τεταρτοκυκλίου.

for i in range(N): # πραγματοποιώ N επαναλήψεις, όσες και τα σημεία
    x = np.random.rand() # τυχαία x συντεταγμένη από 0 έως 1
    y = np.random.rand() # τυχαία y συντεταγμένη από 0 έως 1

    if (x**2 + y**2 < 1): # συνθήκη για να δούμε αν είναι μέσα στο τεταρτοκύκλιο
        n = n + 1 # αν είναι αυξάνουμε το σύνολο των σημειών μέσα στο τεταρτοκύκλιο κατά 1.

my_pi = 4*n/N # εφαρμόζω τη σχέση

print(my_pi)
```
